```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# read csv file
data = pd.read_csv('../benchmark_results/rr_nm_a_tput.csv')
```

```python
# figure 1: line plot of alpha_switch vs avg_load_imbalance
req_rate = 4
data_1 = data[data['req_rate'] == req_rate]
plt.figure(figsize=(10, 6))
sns.lineplot(data=data_1, x='num_models', y='avg_load_imbalance', hue='alpha', marker='o')
plt.yticks(np.arange(0, 0.3, 0.05))
plt.title(f'Average Load Imbalance vs Number of Models (Request Rate = {req_rate} req/s)')
plt.xlabel('Number of Models')
plt.ylabel('Average Load Imbalance')
plt.legend(title='Alpha Switch')
plt.grid(False)
plt.show()
```
